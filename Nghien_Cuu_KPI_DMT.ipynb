# NGHIÊN CỨU & ĐỀ XUẤT MÔ HÌNH KPI GẮN VỚI CÔNG VIỆC TẠI DOANH NGHIỆP

Tài liệu này phân tích thực trạng, khó khăn và đề xuất mô hình đánh giá KPI chuyên sâu, kết hợp giữa Quản lý Công việc (Task-based) và Thẻ điểm cân bằng (Balanced Scorecard - BSC) hoặc OKR để phù hợp với bối cảnh doanh nghiệp.

## 1. Phân Tích Thực Trạng & Khó Khăn Hiện Tại

Dựa trên các dữ liệu và quy trình hiện có (như báo cáo vướng mắc, đánh giá tiến độ): 

- **Đánh giá mang tính Cảm tính / Checklist:** Việc chỉ đánh giá dựa trên trạng thái (Đã hoàn thành / Trễ hạn / Có vướng mắc) dẫn đến tình trạng nhân viên chỉ làm để "đủ số lượng" mà bỏ qua chất lượng thực tế (Outcome).
- **Xử lý Vướng mắc (Bottlenecks):** Nhiều công việc bị kẹt ở trạng thái "Có vướng mắc" nhưng thiếu cơ chế leo thang (escalation) hoặc phân tích nguyên nhân gốc rễ, làm sai lệch kết quả KPI tổng thể.
- **Thiếu Tỷ trọng (Weighting) Rõ ràng:** Một công việc hành chính nhỏ đôi khi được tính ngang bằng với một dự án mang lại doanh thu lớn.
- **Chưa liên kết với Mục tiêu chiến lược:** Các công việc (Tasks) chưa được mapping (gắn kết) rõ ràng vào các góc độ chiến lược của công ty (Ví dụ: BSC - Tài chính, Khách hàng, Vận hành, Phát triển).

## 2. Đề Xuất Mô Hình KPI Phù Hợp: "Hybrid Task-BSC"

Mô hình này không loại bỏ việc giao việc hàng ngày, mà nâng cấp nó lên:
1. **Phân loại Công việc theo 4 khía cạnh BSC:** Mỗi công việc giao xuống phải thuộc 1 trong 4 nhóm (Tài chính, Khách hàng, Quy trình nội bộ, Học tập/Phát triển).
2. **Hệ số Độ khó / Tỷ trọng (Weight):** Công việc A (Tỷ trọng 30%), Công việc B (Tỷ trọng 5%).
3. **Tiêu chí Đánh giá (Rubric):** Không chỉ "Xong/Chưa xong" mà đánh giá theo (Tiến độ - 40%, Chất lượng - 60%).
4. **Cơ chế trừ điểm khi Trễ hạn / Vướng mắc do chủ quan:** Trễ hạn trừ 10% điểm của task đó.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Mock data: Mô phỏng dữ liệu KPI theo mô hình mới
data = {
    'TenCongViec': ['Mở rộng thị trường', 'Bảo trì server', 'Đào tạo nhân sự', 'Cập nhật báo cáo', 'Sửa lỗi phần mềm'],
    'NhomBSC': ['Khách hàng', 'Quy trình', 'Học tập', 'Quy trình', 'Quy trình'],
    'TyTrong(%)': [40, 20, 15, 10, 15],
    'TienDo(%)': [80, 100, 100, 50, 100],
    'DiemChatLuong (1-10)': [8, 9, 7, 5, 10],
    'TrangThai': ['Đang thực hiện', 'Hoàn thành', 'Hoàn thành', 'Trễ hạn', 'Hoàn thành']
}

df = pd.DataFrame(data)
df
```

### 2.1 Thuật toán Tính điểm KPI Tổng hợp
Điểm của mỗi task = Tiêu chí Tiến độ * TyTrong + Tiêu chí Chất lượng * TyTrong
Ở đây ta giả sử: Điểm KPI = (TienDo/100 * 0.4 + DiemChatLuong/10 * 0.6) * TyTrong

```python
def calculate_kpi(row):
    # Tính điểm chuẩn hóa trên 100 cho từng task
    score = (row['TienDo(%)'] / 100.0 * 0.4) + (row['DiemChatLuong (1-10)'] / 10.0 * 0.6)
    
    # Áp dụng hình phạt nếu trễ hạn (giảm 20% số điểm)
    if row['TrangThai'] == 'Trễ hạn':
        score *= 0.8
        
    # Điểm đóng góp vào tổng KPI (tối đa bằng TyTrong)
    return score * row['TyTrong(%)']

df['Diem_KPI_Gop'] = df.apply(calculate_kpi, axis=1)

total_kpi = df['Diem_KPI_Gop'].sum()
print(f"Tổng điểm KPI cuối cùng của nhân viên: {total_kpi:.2f} / 100")

df[['TenCongViec', 'TyTrong(%)', 'TrangThai', 'Diem_KPI_Gop']]
```

## 4. Nghiên cứu Mở rộng: Các Mô hình KPI chuyên biệt cho Quản lý Dự án (Project-based KPIs)

Nếu công ty hoạt động mạnh theo hình thức dự án (phát triển phần mềm, triển khai khách hàng, xây dựng,...), mô hình Task-BSC ở trên có thể chưa phản ánh đủ bức tranh tổng thể về tài nguyên và ngân sách. Dưới đây là 3 mô hình/giải pháp KPI dự án tiên tiến có thể áp dụng hoặc kết hợp:

### 4.1. Mô hình Quản lý Giá trị Thu được (Earned Value Management - EVM)
- **Đặc điểm:** Đây là tiêu chuẩn vàng trong quản trị dự án. Thay vì chỉ hỏi "Đã làm xong việc chưa?" (như Task-based), EVM hỏi "Với số thời gian/tiền đã bỏ ra, chúng ta đã tạo ra được bao nhiêu giá trị thực tế?".
- **Các chỉ số KPI chính (EVM Metrics):**
  - **SPI (Schedule Performance Index):** Chỉ số hiệu suất tiến độ. (SPI > 1: Vượt tiến độ; SPI < 1: Trễ tiến độ).
  - **CPI (Cost Performance Index):** Chỉ số hiệu suất chi phí/nhân công. (CPI > 1: Tiết kiệm; CPI < 1: Vượt ngân sách).
- **Phù hợp cho:** Các dự án lớn có ngân sách hoặc số giờ công (man-hours) được định lượng rõ ràng.

### 4.2. Mô hình KPI Tam Giác Vàng (Time - Cost - Quality/Scope)
- **Đặc điểm:** Đánh giá KPI của thành viên/team dự án dựa trên sự cân bằng của 3 yếu tố khắt khe này.
- **Cách tính điểm KPI:**
  - *Time (Thời gian):* Số ngày hoàn thành thực tế so với Baseline (Kế hoạch).
  - *Cost (Chi phí/Nỗ lực):* Số giờ công đã tiêu tốn so với kế hoạch.
  - *Quality (Chất lượng):* Số lượng lỗi (Bugs/Defects), hoặc tỷ lệ Rework (Làm lại).
- **Phù hợp cho:** Các dự án linh hoạt nhưng có yêu cầu khắt khe về chất lượng bàn giao.

### 4.3. Mô hình Agile / Scrum Metrics (KPI theo chặng - Sprints)
- **Đặc điểm:** Bỏ qua các KPI cồng kềnh, đánh giá team dựa trên tốc độ và khả năng thích ứng.
- **Các chỉ số KPI chính:**
  - **Velocity (Tốc độ):** Số lượng Story Points (Điểm độ khó) team giải quyết được trong 1 tuần/tháng.
  - **Say/Do Ratio:** Tỷ lệ Cam kết / Thực hiện (Cam kết làm 10 việc, làm được 8 việc => 80%).
  - **Lead Time / Cycle Time:** Thời gian từ lúc nhận yêu cầu đến lúc khách hàng dùng được.
- **Phù hợp cho:** Team R&D, Phát triển phần mềm, Marketing - nơi yêu cầu thay đổi liên tục.

```python
# Code mô phỏng đánh giá KPI Dự án theo mô hình EVM (Earned Value Management)
import pandas as pd

project_data = {
    'TenDuAn': ['Dự án ERP', 'App Mobile KH', 'Marketing Q3'],
    'NganSach_GioCong (PV)': [500, 300, 200],      # Planned Value: Số giờ công dự kiến
    'GiaTri_HoanThanh (EV)': [200, 300, 150],      # Earned Value: Giá trị quy đổi thực tế đã làm
    'ThucChi_GioCong (AC)': [250, 280, 200]        # Actual Cost: Số giờ thực tế đã bỏ ra
}

df_project = pd.DataFrame(project_data)

# Tính toán KPI dự án
df_project['SPI (Tiến độ)'] = df_project['GiaTri_HoanThanh (EV)'] / df_project['NganSach_GioCong (PV)']
df_project['CPI (Chi phí)'] = df_project['GiaTri_HoanThanh (EV)'] / df_project['ThucChi_GioCong (AC)']

def danh_gia_du_an(row):
    status = []
    if row['SPI (Tiến độ)'] < 1.0:
        status.append('🔴 Trễ tiến độ')
    else:
        status.append('🟢 Đúng/Vượt tiến độ')
        
    if row['CPI (Chi phí)'] < 1.0:
        status.append('🔴 Vượt ngân sách(Giờ công)')
    else:
        status.append('🟢 Tiết kiệm(Giờ công)')
    return ' | '.join(status)

df_project['DanhGia_KPI'] = df_project.apply(danh_gia_du_an, axis=1)
df_project
```

## 5. Phân tích KPI chuyên biệt cho Ngành Xây dựng, Kỹ thuật & Giải tỏa Đền bù

Với đặc thù dự án xây dựng, giải phóng mặt bằng (GPMB) và kỹ thuật hạ tầng, dự án thường có ngân sách lớn (CAPEX), tiến độ phụ thuộc nhiều vào yếu tố bên ngoài (pháp lý, chính quyền, người dân) và phải tuân thủ nghiêm ngặt theo các giai đoạn (Milestones). Việc áp dụng KPI cần tập trung vào **Quản trị Rủi ro (Risk Management)**, **Pháp lý/GPMB** và **Hiệu quả thi công** thay vì các chỉ số Agile/Khách hàng.

### 5.1. Nhóm KPI Quản lý Giải phóng mặt bằng (GPMB) & Pháp lý
Khâu này thường xuyên bị "Có vướng mắc". KPI không nên chỉ là "Đã xong thủ tục chưa?" mà phải lượng hóa:
- **Tỷ lệ hộ dân ký biên bản bàn giao / Tổng số hộ (%):** Đo lường trực tiếp hiệu quả của team đền bù.
- **Tiến độ giải ngân vốn đền bù:** Tỷ lệ % số tiền đã chi trả thành công so với ngân sách GPMB được duyệt.
- **Thời gian trung bình xử lý khiếu nại (Lead time of grievance resolution):** Số ngày từ lúc nhận đơn thư đến lúc giải quyết xong.
- **Tỷ lệ hồ sơ pháp lý được phê duyệt đúng hạn (Right-First-Time):** Đánh giá năng lực của ban pháp lý/dự án khi làm việc với cơ quan nhà nước.

### 5.2. Nhóm KPI Quản lý Thi công & Kỹ thuật (Áp dụng EVM Khối lượng)
- **Tỷ lệ hoàn thành khối lượng (Physical Progress %):** Số lượng cọc đã ép, khối lượng bê tông đã đổ, hoặc số km đường đã thảm nhựa.
- **Chỉ số Hiệu suất Tiến độ (SPI) và Chi phí (CPI) trong EVM:**
  - *Ví dụ:* Kế hoạch tháng này phải tiêu hết 10 tỷ (PV) và hoàn thành 20% móng. Thực tế làm được 15% móng (EV = 7.5 tỷ) nhưng tiêu hết 9 tỷ (AC) => Trễ tiến độ (SPI < 1) và Vượt chi phí (CPI < 1).
- **Tỷ lệ Rework (Làm lại) / Sai sót kỹ thuật:** Khối lượng phải đập đi làm lại do lỗi kỹ thuật tư vấn/thi công.
- **An toàn Lao động (HSE):** Số ngày không xảy ra tai nạn lao động (Zero Lost Time Injury).

### 5.3. Xử lý Trạng thái "Có vướng mắc" trong Xây dựng/GPMB
- Trong xây dựng, vướng mắc (mưa bão, dân không nhận tiền) là chuyện bình thường. Do đó, hệ thống KPI cần có **Tiêu chí loại trừ (Excusable Delays)**.
- **Quy tắc:** Nếu vướng mắc do khách quan (Thiên tai, Chính quyền chậm duyệt) => Không trừ điểm KPI tiến độ của cá nhân, nhưng yêu cầu KPI về **"Tốc độ báo cáo và đề xuất phương án giải quyết"**.
- Nếu vướng mắc do chủ quan (quên nộp hồ sơ, thi công sai bản vẽ) => Trừ 30-50% điểm KPI của hạng mục đó.
